```python
# ۴. دانلود ویدیوی خروجی
import glob
from google.colab import files

output_videos = glob.glob('/content/CodeFormer/results/**/*.mp4', recursive=True)
if output_videos:
    latest_output = max(output_videos, key=os.path.getctime)
    print(f"در حال دانلود فایل خروجی: {latest_output}")
    files.download(latest_output)
else:
    print("فایل ویدیویی در پوشه results یافت نشد. لطفاً تب فایل‌ها در سمت چپ را چک کنید.")
```
